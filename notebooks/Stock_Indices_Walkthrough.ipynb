## All available

Fetch each index as a frame; report count + a few ticker/name pairs. Needs network.

````python
import pandas as pd

rows = []
for name in si.available:
    frame = si.get(name, as_frame=True)
    if frame.empty:
        sample = "(failed)"
    else:
        sample = "; ".join(
            f"{r.ticker} ({r.name})" for r in frame.head(3).itertuples()
        )
    rows.append({
        "name": name,
        "label": si.label(name),
        "n": len(frame),
        "sample": sample,
    })
    print(f"{name:12} {len(frame):4d}  {si.label(name)}")

summary = pd.DataFrame(rows)
summary
````
