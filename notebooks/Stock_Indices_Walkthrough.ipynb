## All available

Fetch each index, report count + a few sample tickers. Live Wikipedia scrape — needs network.

````python
rows = []
for name in si.available:
    tickers = si.get(name)
    rows.append({
        "name": name,
        "label": si.label(name),
        "n": len(tickers),
        "sample": ", ".join(tickers[:5]) if tickers else "(failed)",
    })
    print(f"{name:12} {len(tickers):4d}  {si.label(name)}")

import pandas as pd

summary = pd.DataFrame(rows)
summary
````
